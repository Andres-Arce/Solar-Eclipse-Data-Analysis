##  Load File: `solar-eclipses.csv`

````python
import pandas as pd
# from utils.fontStyle import main

pd.set_option('display.max_colwidth', 1000)
df = pd.read_csv('../data/solar-eclipses-clean.csv')
````

````python
"""
Cuando tuvo lugar el eclipse solar mas largo y el eclipse total mas largo?
"""
eclipse_longer = df.loc[df["duration_total"].idxmax()]

eclipses_total = df.query("eclipse_type == 'Total'")
eclipses_total_longer = eclipses_total.loc[eclipses_total["duration_total"].idxmax()]

print(
f"""
    El lugar donde tuvo el eclipse solar mas largo fue en {eclipse_longer["region"]} con un duracion
    total de {eclipse_longer["duration_total"]} 

    El lugar que tuvo el eclipse total mas largo fue en {eclipses_total_longer["region"]} con una duracion
    total de {eclipses_total_longer["duration_total"]}
"""
)

````

````python
# Cuales son la duracion media de los eclipses solares totales
df.groupby("eclipse_type")["duration_total"].mean()
````

````python
# Cual tipo de eclipse es mas frecuente?
df["eclipse_type"].value_counts()
````

````python
# Cuales seran los proximos 10 eclipses solares
df.query("year >= 2026").head(10)
````

````python
#Cual fue el lugar donde el eclipse duro menos tiempo
shortest_eclipse = df.loc[df["duration_total"].idxmin()]

print(
    f"The shortest eclipse occurred in {shortest_eclipse['region']} "
    f"with a duration of {shortest_eclipse['duration_total']} seconds."
)
````

````python
# ¿Cual es el lugar donde mas frecuente pasa el eclipse y cual el menos frecuente?
place_more_frecuency = df.loc[df["region"].value_counts()].max()
place_minor_frecuency = df.loc[df["region"].value_counts()].min()

print(
    f"The place with more frequency is {place_more_frecuency["region"]} "
    f"and the minor frecuency is {place_minor_frecuency["region"]}"
)
````

````python
# cual es la magnitud mas grande y mas pequeña regisrada
magnitude_greater = df["magnitude"].idxmax()
magnitude_minor = df["magnitude"].idxmin()

print(magnitude_greater)
print(magnitude_minor)
````

````python
# Cual es el año con mas eclipses?
year_more_eclipses = df["year"].value_counts().idxmax()
year_minor_eclipses = df["year"].value_counts().idxmin()

print(year_more_eclipses)
print(year_minor_eclipses)
````
